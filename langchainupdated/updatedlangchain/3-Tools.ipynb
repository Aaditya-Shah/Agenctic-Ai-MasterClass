````python
import os
from dotenv import load_dotenv, find_dotenv # <--- ADD THIS to unlock your file
from langchain.chat_models import init_chat_model

load_dotenv(find_dotenv())

model = init_chat_model(
"groq:qwen/qwen3-32b",
max_tokens=1000,
temperature=0.0
)
messages = [

    ("system", "You are genius expolit developer, you can find vulnerabilities in code and write exploits for them. You are also a helpful assistant who can answer questions about code and programming. You will be given a code snippet, and you will analyze it for vulnerabilities. If you find any vulnerabilities, you will write an exploit for them. If you do not find any vulnerabilities, you will explain why the code is secure. You will also answer any questions about the code and programming in general."),
    ("user", "Here is the code snippet: \n\n```python\nimport os\nimport subprocess\n\ndef run_command(command):\n    os.system(command)\n\nif __name__ == \"__main__\":\n    user_input = input(\"Enter a command to run: \")\n    run_command(user_input)\n```")
 ]
response = model.invoke(messages)
print(response)
````

````python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the current weather for a given location. can u read this """
#MOCK WEATHER DEMO
    return f"It's sunny in {location}AA"

model_with_tools=model.bind_tools([get_weather])

````

````python
response = model_with_tools.invoke("WHATS THE WEATHER IN BIRATNAAGT")
#print(response)
for tool_call in response.tool_calls:
    #VIEW TOOL CALLS MADE BY THE MODEL
    print(f"Tool : {tool_call['name']}")
    print(f"Arguments : {tool_call['args']}")
    #print(f"Content: {additional_kwargs['reasoning_content']}")
    print(f"Reasoning: {response}")
````
