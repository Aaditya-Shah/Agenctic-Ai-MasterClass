````python
import os
from langchain.chat_models import init_chat_model
from langchain_core.tools import tool 

# 1. Provide a dummy API key to bypass initialization checks
os.environ["OPENAI_API_KEY"] = "lm-studio"

# 2. Initialize pointing directly to your local LM Studio server
model = init_chat_model(
    model="qwen/qwen3-1.7b", 
    model_provider="openai",                    
    base_url="http://192.168.10.103:1234/v1",  
    temperature=0.0
)

# 3. Define your tool
@tool
def get_weather(location: str) -> str:
    """Get the current weather for a given location."""
    return f"The weather there is not good and filled with clouds in {location}AA"

# 4. Bind the tool to the model
model_with_tools = model.bind_tools([get_weather])

print("Sending request to LM Studio... Please wait.")

# 5. Invoke the model
response = model_with_tools.invoke("WHATS THE WEATHER IN BIRATNAGAR?")

# 6. View the Tool Calls
print("\n--- Tool Selection ---")
for tool_call in response.tool_calls:
    print(f"Tool Name: {tool_call['name']}")
    print(f"Arguments: {tool_call['args']}\n")

# 7. Fallback to view raw text if the model failed to use the tool
print("--- Raw Content (If tool calls are empty) ---")
print(response.content)
````

### Tool Execution Loops  

````python
messages = [{"role": "user", "content": "What is the weather in Biratnagar which fruit is suitable? its 84 degree here and not cloudy its sunny"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

#execute tools and collenct results
for tool_call in ai_msg.tool_calls:
    #execute the tools with the generated arguments
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

#pass results back to the model for final response
final_response = model_with_tools.invoke(messages)
print(final_response.content)
````
