### USING B.AI TOKENS 500K


```python
import json
import requests

# 1. Ensure stream=True is passed as a parameter to requests.post
response = requests.post(
    "https://api.b.ai/v1/chat/completions",
    headers={
        "Authorization": "sk-14mf0nlreh9tez28tx7g2op1tzwawjpq",
        "Content-Type": "application/json",
    },
    json={
        "model": "deepseek-v4-flash",
        "messages": [
            {"role": "user", "content": "Write essay about Hacking"}
        ],
        "stream": True,  # Tells the API provider to stream chunks
        "temperature": 0.7,
        "max_tokens": 1000,
    },
    stream=True,  # CRITICAL: Tells the Python requests library to keep the connection open
)

# 2. Iterate over the stream line-by-line
for line in response.iter_lines():
    if line:
        # Decode the raw bytes into a string
        decoded_line = line.decode('utf-8')
        
        # Stream lines in OpenAI-compatible APIs always start with "data: "
        if decoded_line.startswith("data: "):
            content = decoded_line[6:].strip()
            
            # Check for the stream termination signal
            if content == "[DONE]":
                break
                
            try:
                # Parse the raw JSON string chunk
                chunk_json = json.loads(content)
                
                # Safely extract the text delta
                delta = chunk_json['choices'][0]['delta'].get('content', '')
                
                # Print each token chunk instantly without line breaks
                print(delta, end="", flush=True)
            except json.JSONDecodeError:
                pass

print("\n\n--- Stream Completed ---")
```
