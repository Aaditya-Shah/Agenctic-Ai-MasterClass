```python
import os
from dotenv import find_dotenv, load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

# 1. Load the API key
load_dotenv(find_dotenv())
my_api_key = os.getenv("GOOGLE_API_KEY") or os.getenv("GEMINI_API_KEY")

# 2. Initialize the model (using the correct 1.5 version)
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash", 
    api_key=my_api_key
)

# 3. Invoke the model directly! (No complicated agent wrapper needed)
response = llm.invoke("Hello! what is 2+2")

# 4. Print the result
print(response.text)
```
