#### message metadata

```python
human_msg = HumanMessage(
    
    name="alice", #identify user
    id="msg_123", #for tracing
    content="Hello! my name is {['name']} "
)
response = model.invoke([human_msg])
print(response.content)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
ai_msg = AIMessage("I can help you with that! shit")
messages = [
    SystemMessage("You are a expert Duffer of Mathematics and uses chemistry for physics and does math in binary to catch Criminal"),
    HumanMessage("Write a desrciption to trap a criminal from murder charge"),
    ai_msg,
    HumanMessage("Greate! WHats integration of 200x from 0 to 100"),
    HumanMessage("WHO AM I WHATS MY NAME ETC"),
    HumanMessage(
    content="Hello!",
    name="alice", #identify user
    id="msg_123", #for tracing
    prompt = "WHO AM I"
)
]

response=model.invoke(messages)
print(response.content)
```

```python
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"] = "lm-s"

model = init_chat_model(
    model="qwen/qwen3-1.7b",
    model_provider="openai",                    
    base_url="http://192.168.10.103:1234/v1",  
    temperature=0.0
)

#define template with {username}

prompt_template = ChatPromptTemplate.from_messages([

    ("system", "You are genius"),
    ("user", "HELLO !! my name is {username}")

])

formatted_prompt = prompt_template.invoke({"username": "Alice"})
response = model.invoke(formatted_prompt)
print("AI RESPONSE")
print(response.content)
```
